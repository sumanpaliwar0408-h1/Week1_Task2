## 1. Data Preparation and Validation

Before calculating business metrics, the dataset was inspected and prepared for analysis.

The following checks and transformations were performed:

- Column names were standardized to lowercase `snake_case`.
- The `sub-category` field was standardized to `sub_category`.
- Fully duplicated rows were checked and none were found.
- Duplicate `row_id` values were checked and none were found.
- `order_date` and `ship_date` were converted to datetime format.
- Date conversion was validated and no invalid order or shipping dates were detected.
- `year`, `month`, and `year_month` fields were created from `order_date` for time-based analysis.

Repeated `order_id` values were retained because the dataset is stored at the order-line level, meaning one valid order can contain multiple product rows.

## 2. Order-Level Analysis Logic

The source dataset contains one row per order line rather than one row per complete order. Therefore, an order-level summary was created by grouping records using `order_id`.

For each unique order, the analysis aggregates:

- Total order revenue
- Total units purchased
- Customer ID
- Order date

This distinction is important because metrics such as **Total Orders** and **Average Order Value** must be calculated at the unique-order level rather than by counting individual product rows.

```python
# Creating an order-level summary

order_summary = (
    df.groupby('order_id')
      .agg(
          order_revenue=('sales', 'sum'),
          total_units=('quantity', 'sum'),
          customer_id=('customer_id', 'first'),
          order_date=('order_date', 'first')
      )
      .reset_index()
)

order_summary.head()
```

## 3. Core KPI Analysis

Six core KPIs were calculated to summarize overall sales performance.

### KPI Definitions

- **Total Revenue:** Sum of sales across all order line items.
- **Total Orders:** Number of unique `order_id` values.
- **Average Order Value (AOV):** Total Revenue divided by Total Orders.
- **Total Units Sold:** Sum of the `quantity` field.
- **Top-Selling Category:** Product category contributing the highest total revenue.
- **Top Revenue-Generating Product:** Individual product generating the highest total sales revenue.

```python
total_revenue = df['sales'].sum()

print(f"Total Revenue: ${total_revenue:,.2f}")
```

```python
total_orders = df['order_id'].nunique()

print(f"Total Orders: {total_orders:,}")
```

```python
average_order_value = total_revenue / total_orders

print(f"Average Order Value: ${average_order_value:,.2f}")
```

```python
total_units_sold = df['quantity'].sum()

print(f"Total Units Sold: {total_units_sold:,}")
```

```python
category_sales = (df.groupby('category')['sales'].sum().sort_values(ascending=False))
category_sales
```

```python
top_category = category_sales.idxmax()
top_category_revenue = category_sales.max()

print("Top_Selling_Category:", top_category)
print(f"Category_Revenue: ${top_category_revenue:,.2f}")
```

```python
product_sales = (df.groupby('product_name')['sales'].sum().sort_values(ascending=False))
product_sales.head(10)
```

```python
top_product = product_sales.idxmax()
top_product_revenue = product_sales.max()

print("Top Revenue-Generating Product:", top_product)
print(f"Product Revenue: ${top_product_revenue:,.2f}")
```

```python
#KPI summary Tbale-
kpi_summary = pd.DataFrame({
    'KPI': [
        'Total Revenue',
        'Total Orders',
        'Average Order Value',
        'Total Units Sold',
        'Top-Selling Category',
        'Top Revenue-Generating Product'
    ],
    'Value': [
        f"${total_revenue:,.2f}",
        f"{total_orders:,}",
        f"${average_order_value:,.2f}",
        f"{total_units_sold:,}",
        top_category,
        top_product
    ]
})

kpi_summary
```

## 4. Sales Trend Analysis

Sales performance was analyzed over time using yearly and monthly aggregations of revenue, unique orders, and units sold.

### Yearly Performance

Revenue changed as follows:

- **2014:** $484,247.50
- **2015:** $470,532.51
- **2016:** $609,205.60
- **2017:** $733,215.26

Revenue decreased by approximately **2.83% in 2015**, followed by strong increases of approximately **29.47% in 2016** and **20.36% in 2017**.

The number of unique orders also increased from **969 in 2014** to **1,687 in 2017**, while units sold increased from **7,581** to **12,476**.

### Trend Interpretation

The results show that sales performance was not consistently increasing throughout the entire period because revenue declined slightly in 2015. However, the subsequent growth in both 2016 and 2017, together with increases in order count and units sold, indicates substantially stronger sales activity during the later years of the dataset.

The dataset alone does not identify the cause of this growth, so factors such as marketing activity, market expansion, or seasonal demand should not be presented as confirmed explanations.

```python
yearly_sales = (
    df.groupby('year')
      .agg(
          total_revenue=('sales', 'sum'),
          total_orders=('order_id', 'nunique'),
          total_units=('quantity', 'sum')
      )
      .reset_index()
)

yearly_sales['revenue_growth_pct'] = (
    yearly_sales['total_revenue']
    .pct_change()
    .mul(100)
    .round(2)
)

yearly_sales
```

```python
monthly_sales = (
    df.groupby('year_month')
      .agg(
          total_revenue=('sales', 'sum'),
          total_orders=('order_id', 'nunique'),
          total_units=('quantity', 'sum')
      )
      .reset_index()
)

monthly_sales.head()
```

```python
monthly_sales['revenue_growth_pct'] = (
    monthly_sales['total_revenue']
    .pct_change()
    .mul(100)
    .round(2)
)

monthly_sales.head(12)
```

```python
highest_sales_month = monthly_sales.loc[
    monthly_sales['total_revenue'].idxmax()
]

highest_sales_month
```

```python
lowest_sales_month = monthly_sales.loc[
    monthly_sales['total_revenue'].idxmin()
]

lowest_sales_month
```

### Monthly Sales Performance

The highest-revenue month in the dataset was **November 2017**, generating approximately **$118,447.82** from **261 orders** and **1,840 units sold**.

The lowest-revenue month was **February 2014**, with approximately **$4,519.89** in revenue from **28 orders** and **159 units sold**.

These results demonstrate substantial variation in monthly sales performance. November 2017 represents the strongest observed monthly performance, but the dataset alone is insufficient to attribute the peak to a specific seasonal, promotional, or external cause.

```python
monthly_pattern = (
    df.groupby(df['order_date'].dt.month)
      .agg(
          total_revenue=('sales', 'sum'),
          total_orders=('order_id', 'nunique'),
          total_units=('quantity', 'sum')
      )
      .reset_index()
)

monthly_pattern['month_name'] = pd.to_datetime(
    monthly_pattern['order_date'],
    format='%m'
).dt.month_name()

monthly_pattern
```

## 5. Category and Product Performance

Product performance was evaluated using both revenue and unit volume.

### Category Performance

Technology generated the highest category revenue at approximately **$836,154.03**, representing **36.40%** of total revenue.

Furniture generated approximately **$741,999.80** (**32.30%**), while Office Supplies generated approximately **$719,047.03** (**31.30%**).

Although Technology was the largest revenue contributor, revenue was distributed relatively closely across all three major product categories.

### Product Performance

The **Canon imageCLASS 2200 Advanced Copier** was the highest-revenue product, generating approximately **$61,599.82** from **20 units sold**.

In contrast, **Staples** was the highest-volume product with **215 units sold**.

Therefore, the highest-revenue product and highest-volume product were not the same. This illustrates why both revenue and unit volume should be considered when evaluating product performance.

```python
category_performance = (
    df.groupby('category')
      .agg(
          total_revenue=('sales', 'sum'),
          total_orders=('order_id', 'nunique'),
          total_units=('quantity', 'sum'),
          total_profit=('profit', 'sum')
      )
      .sort_values('total_revenue', ascending=False)
)

category_performance['revenue_share_pct'] = (
    category_performance['total_revenue']
    / total_revenue * 100
).round(2)

category_performance
```

```python
product_performance = (
    df.groupby('product_name')
      .agg(
          total_revenue=('sales', 'sum'),
          total_units=('quantity', 'sum'),
          total_orders=('order_id', 'nunique'),
          total_profit=('profit', 'sum')
      )
      .sort_values('total_revenue', ascending=False)
)

product_performance.head(10)
```

```python
top_products_by_units = (
    df.groupby('product_name')
      .agg(
          total_units=('quantity', 'sum'),
          total_revenue=('sales', 'sum')
      )
      .sort_values(
          ['total_units', 'total_revenue'],
          ascending=False
      )
)

top_products_by_units.head(10)
```

```python
category_yearly_sales = (
    df.pivot_table(
        index='year',
        columns='category',
        values='sales',
        aggfunc='sum'
    )
    .round(2)
)

category_yearly_sales
```

## 6. Regional Performance

Sales performance was compared across geographic regions using revenue, unique orders, customers, units sold, profit, and average order value.

The regional revenue ranking was:

1. **West:** $725,457.82 — 31.58% of total revenue
2. **East:** $678,781.24 — 29.55%
3. **Central:** $501,239.89 — 21.82%
4. **South:** $391,721.90 — 17.05%

The **West** was therefore the highest-revenue region, generating approximately **$725,457.82** from **1,611 unique orders**.

However, the East recorded a higher average order value than the West. This demonstrates that the region generating the most total revenue does not necessarily have the highest average value per order.

```python
region_performance = (
    df.groupby('region')
      .agg(
          total_revenue=('sales', 'sum'),
          total_orders=('order_id', 'nunique'),
          unique_customers=('customer_id', 'nunique'),
          total_units=('quantity', 'sum'),
          total_profit=('profit', 'sum')
      )
      .sort_values('total_revenue', ascending=False)
)

region_performance['average_order_value'] = (
    region_performance['total_revenue']
    / region_performance['total_orders']
)

region_performance['revenue_share_pct'] = (
    region_performance['total_revenue']
    / total_revenue * 100
).round(2)

region_performance.round(2)
```

```python
state_performance = (
    df.groupby('state')
      .agg(
          total_revenue=('sales', 'sum'),
          total_orders=('order_id', 'nunique'),
          total_units=('quantity', 'sum'),
          total_profit=('profit', 'sum')
      )
      .sort_values('total_revenue', ascending=False)
)

state_performance.head(10).round(2)
```

```python
city_performance = (
    df.groupby(['state', 'city'])
      .agg(
          total_revenue=('sales', 'sum'),
          total_orders=('order_id', 'nunique'),
          unique_customers=('customer_id', 'nunique'),
          total_units=('quantity', 'sum')
      )
      .sort_values('total_revenue', ascending=False)
)

city_performance.head(10).round(2)
```

```python
top_region = region_performance.index[0]

print("Top Revenue-Generating Region:", top_region)
print(
    f"Revenue: ${region_performance.loc[top_region, 'total_revenue']:,.2f}"
)
```

## 7. Customer Insights

Customer-level performance was analyzed using revenue, unique order count, units purchased, profit, and average order value.

The dataset contains **793 unique customers**.

### Highest-Value and Most Frequent Customers

**Sean Miller** generated the highest total customer revenue:

- Revenue: **$25,043.05**
- Orders: **5**

The most frequent customer was **Emily Phan**:

- Orders: **17**
- Revenue: **$5,478.06**

The highest-revenue customer was therefore not the customer who placed the most orders. This suggests that customer value should be evaluated using both purchase frequency and monetary contribution rather than order frequency alone.

```python
customer_performance = (
    df.groupby('customer_id')
      .agg(
          customer_name=('customer_name', 'first'),
          total_revenue=('sales', 'sum'),
          total_orders=('order_id', 'nunique'),
          total_units=('quantity', 'sum'),
          total_profit=('profit', 'sum')
      )
)

customer_performance['average_order_value'] = (
    customer_performance['total_revenue']
    / customer_performance['total_orders']
)

customer_performance.head()
```

```python
top_customers_by_revenue = (
    customer_performance
    .sort_values('total_revenue', ascending=False)
)

top_customers_by_revenue.head(10).round(2)
```

```python
top_customers_by_orders = (
    customer_performance
    .sort_values(
        ['total_orders', 'total_revenue'],
        ascending=False
    )
)

top_customers_by_orders.head(10).round(2)
```

```python
repeat_customers = customer_performance[
    customer_performance['total_orders'] > 1
]

one_time_customers = customer_performance[
    customer_performance['total_orders'] == 1
]

print("Repeat customers:", len(repeat_customers))
print("One-time customers:", len(one_time_customers))
```

```python
repeat_customer_pct = (
    len(repeat_customers)
    / len(customer_performance)
    * 100
)

print(f"Repeat Customer Percentage: {repeat_customer_pct:.2f}%")
```

```python
repeat_customer_revenue = repeat_customers['total_revenue'].sum()

repeat_customer_revenue_share = (
    repeat_customer_revenue
    / total_revenue
    * 100
)

print(f"Repeat Customer Revenue: ${repeat_customer_revenue:,.2f}")
print(
    f"Repeat Customer Revenue Share: "
    f"{repeat_customer_revenue_share:.2f}%"
)
```

### Repeat Customer Analysis

Customers with more than one unique order were classified as repeat customers.

The analysis identified:

- Total customers: **793**
- Repeat customers: **781**
- One-time customers: **12**
- Repeat customer rate: **98.49%**
- Repeat customer revenue share: **99.78%**

Repeat customers account for nearly all revenue observed in this dataset. This shows that the historical sales represented in the dataset are overwhelmingly associated with customers who placed multiple orders.

This result describes the purchasing pattern in the available dataset and should not, by itself, be interpreted as evidence that repeat purchasing caused higher revenue.

```python
segment_performance = (
    df.groupby('segment')
      .agg(
          total_revenue=('sales', 'sum'),
          total_orders=('order_id', 'nunique'),
          unique_customers=('customer_id', 'nunique'),
          total_units=('quantity', 'sum'),
          total_profit=('profit', 'sum')
      )
      .sort_values('total_revenue', ascending=False)
)

segment_performance['average_order_value'] = (
    segment_performance['total_revenue']
    / segment_performance['total_orders']
)

segment_performance['revenue_share_pct'] = (
    segment_performance['total_revenue']
    / total_revenue
    * 100
).round(2)

segment_performance.round(2)
```

## 10. Analysis Limitations and Assumptions

The following limitations and analytical assumptions should be considered when interpreting the results:

- The dataset is stored at the **order-line level**, so unique `order_id` values are used when calculating order-based metrics.
- Revenue is calculated from the provided `sales` field.
- The analysis describes patterns contained in the dataset and does not establish causal explanations for changes in sales.
- Monthly and yearly differences may reflect multiple factors that are not represented in the available variables.
- Repeat customers are defined as customers with more than one unique order in the dataset.
- The analysis is limited to the time period and customers represented in the supplied Sample Superstore dataset.

These assumptions ensure that observations remain supported by the available data and are not interpreted beyond what the dataset can demonstrate.
